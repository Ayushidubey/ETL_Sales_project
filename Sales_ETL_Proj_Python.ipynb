Transform data 

```python
sales_df = pd.read_csv("Skillup_dataset/Spring_dataset/LA_Retail_Sales.csv")
#sales_df.info()
## convert column name to consistent column name replace space with _ and lowercase
sales_df.columns = sales_df.columns.str.lower().str.replace(' ','_')
sales_df.info()
```

```python
#missing_unit_indices = sales_df.query("unit_sales.isna()").index
#print(missing_unit_indices)
sales_df["unit_sales"] =  sales_df["unit_sales"].fillna(0)
sales_df["dollar_sales"] = sales_df["dollar_sales"].fillna(0)
```

```python
sales_df = sales_df.dropna(subset=["unit_sales","dollar_sales"])
sales_df.info()
```

```python
#add new column in Sales_df for revenue

sales_df["rev_per_unit"] = sales_df["dollar_sales"]/sales_df["unit_sales"]
sales_df.info()
```

```python
#to fetch distinct store_zip and count

sales_df.store_zip.value_counts()

#different store_zi we got but there is bad zip 900XX  will need to replace XX with 00 make good record
```

```python
#different store_zip we got but there is bad zip 900XX  will need to replace XX with 00 make good record
sales_df["store_zip"] = sales_df["store_zip"].str.replace("X","0")
```

```python
#cleanup data type for all the columns where datatype is misinterpreted or wrong like date column should have date datatype but its string
sales_df["date"] = pd.to_datetime(sales_df["date"],format ='mixed')
sales_df["promotion_flag"] = sales_df["promotion_flag"].astype(bool)
sales_df["unit_sales"] =sales_df["unit_sales"].astype(int)
```
